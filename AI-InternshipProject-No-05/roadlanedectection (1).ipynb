```python
# ===============================
# STEP 1: Install dependencies
# ===============================
!pip install ultralytics opencv-python matplotlib

import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

# ===============================
# STEP 2: Load pretrained YOLOv8 (lane-detection trained weights)
# Note: You need a model trained on lanes (e.g. TuSimple dataset)
# Example: "best.pt" after training, otherwise will fail with coco weights
# ===============================
# For now using yolov8n-seg pretrained (COCO), replace with lane-trained weights
model = YOLO("yolov8n-seg.pt")

# ===============================
# STEP 3: Input video or image
# ===============================
video_path = "/content/road.mp4"   # apna road video yahan daalo
cap = cv2.VideoCapture(video_path)

out = cv2.VideoWriter(
    "lane_output.mp4", cv2.VideoWriter_fourcc(*"mp4v"),
    cap.get(cv2.CAP_PROP_FPS),
    (int(cap.get(3)), int(cap.get(4)))
)

# ===============================
# STEP 4: Run YOLO inference frame by frame
# ===============================
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # YOLO inference (segmentation mode for lanes)
    results = model.predict(frame, conf=0.3)

    # Annotated frame
    annotated_frame = results[0].plot()

    # Show frame (optional for Colab)
    # cv2_imshow(annotated_frame)  # for Google Colab
    out.write(annotated_frame)

cap.release()
out.release()

print("✅ Lane detection video saved as lane_output.mp4")
```
